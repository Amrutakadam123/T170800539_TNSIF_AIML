```python
df = pd.read_csv('data.csv')
print(df.head())
```

```python
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])
```

```python
print(df.shape)
```

```python
print(df.dtypes)
```

```python
print(df.isnull().sum())
```

```python
print("Total Missing Values:", df.isnull().sum().sum())
```

```python
numeric_columns = df.select_dtypes(include='number').columns

print(numeric_columns)
```

```python
for column in numeric_columns:
    df[column] = df[column].fillna(df[column].mean())

print(df)
```

```python
numeric_columns = df.select_dtypes(include='number').columns

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())

print(df)
```

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
```

```python
categorical_columns = df.select_dtypes(include='object').columns

print(categorical_columns)
```

```python
for column in categorical_columns:
    df[column] = df[column].fillna(df[column].mode()[0])

print(df)
```

```python
print("Duplicate Rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print(df)
```

```python
df["Admission_Date"] = pd.to_datetime(
    df["Admission_Date"],
    dayfirst=True,
    format='mixed'
)

print(df["Admission_Date"])
```

```python
df["Date_of_Birth"] = pd.to_datetime(
    df["Date_of_Birth"],
    dayfirst=True,
    format='mixed',
    errors='coerce'
)

print(df["Date_of_Birth"])
```

```python
df.columns = df.columns.str.lower()

print(df.columns)
```

```python
print(df.select_dtypes(include='float').columns)
```

```python
df["age"] = df["age"].astype(int)
```

```python
df["math_marks"] = df["math_marks"].astype(int)
```

```python
columns = ["math_marks", "science_marks", "english_marks"]

df[columns] = df[columns].fillna(
    df[columns].mean()
)

print(df[columns])
```

```python
df["math_marks"] = df["math_marks"] * 1.10

print(df["math_marks"])
```

```python
columns = [
    "math_marks",
    "science_marks",
    "english_marks"
]

df[columns] = df[columns] * 1.10

print(df[columns])
```

```python

df["department"] = df["department"].str.lower()

print(df["department"].unique())
```

```python
df["department"] = df["department"].str.strip()
```

```python
df["average_marks"] = (
    df["math_marks"] +
    df["science_marks"] +
    df["english_marks"]
) / 3

df["pass_fail"] = df["average_marks"].apply(
    lambda x: "Pass" if x >= 40 else "Fail"
)

print(df[["name", "average_marks", "pass_fail"]])
```

```python
students = df[df["math_marks"] > 80]

print(students)
```

```python
df.to_csv(
    "student_preprocessed.csv",
    index=False
)

print("Dataset saved successfully!")
```
