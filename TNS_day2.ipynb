```python
class Dog:
  species = "Canine" #Class Attribute

  def __init__(self, name, age):
      self.name = name  #Instance Attribute
      self.age = age

dog1 = Dog("Buddy",3)
dog2 = Dog("Miky",7)

print(dog1.name, dog1.age)
print(dog2.name, dog2.age)
```
