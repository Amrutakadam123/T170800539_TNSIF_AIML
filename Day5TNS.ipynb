```python
# Data PreProcessing in Python
import pandas as pd
import numpy as np
```

```python
data = {
    "Name": ["Ram", "Shyam", "Ram", "shyam", "Hari", "None"],
    "Age": [20, 30, 20, 30, 60, None],
    'Salary': [25000, 30000, 28000, np.nan, 25000, 100000],
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male', 'Male'],
    "City": ["Kathmandu", "Bhaktapur", "Lalitpur", "Bhaktapur", "Kathmandu", None]


}
df = pd.DataFrame(data)
print(df)
```

```python
from sklearn.preprocessing import StandardScaler

# Check missing values
print(df.isnull().sum())

# Fill missing values
df['Name'] = df['Name'].fillna('Unknown')
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Salary'] = df['Salary'].fillna(df['Salary'].median())

# Remove Duplicates
df.drop_duplicates(inplace=True)

# convert categorical data
df['Gender']=df['Gender'].map({
    'Male':1,
    'Female':0
})
# one hot encoding
df = pd.get_dummies(df,columns=['City'])

#Display Preprocessing data
print(df)

#Feature Scaling
scaler = StandardScaler()

df[['Age','Salary']] = scaler.fit_transform(
    df[['Age', 'Salary']])
#Display final data
print(df)
```

```python
#Increse values by 10%
df['Salary'] = df['Salary'] * 1.1
print(df)
```
